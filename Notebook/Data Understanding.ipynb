```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
train_df = pd.read_csv("DailyDelhiClimateTrain.csv")
test_df = pd.read_csv("DailyDelhiClimateTest.csv")
```

```python
print("Train Duplicates :", train_df.duplicated().sum())
print("Test Duplicates :", test_df.duplicated().sum())
```

```python
df = pd.concat([train_df, test_df], ignore_index=True)
print("Merged Dataset Shape :", df.shape)
```

```python
df["date"] = pd.to_datetime(df["date"])
df.info()
```

The date column is converted into datetime format for time-series analysis.

```python
print("Start Date :", df["date"].min())
print("End Date   :", df["date"].max())
```

```python
df.to_csv("DailyDelhiClimateMerged.csv", index=False)
print("Merged dataset saved successfully.")
```
